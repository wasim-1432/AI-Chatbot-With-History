```python
from dotenv import load_dotenv
load_dotenv()
from google import genai

client = genai.Client()

interaction = client.interactions.create(
    model="gemini-3.6-flash",
    input="Tell me who invested samosa in 3 lines."
)

print(interaction.output_text)
```

```python
def generate_recepie(ingredients, cuisine, diet):
    ''' 
        Generate a recepie using ingredients provided.
        Cuisine can be "Italian" "Maxican", "Indian", "Any", etc.
        Diet can be "Vegan", "Vegitarian","Jain","Any", etc.
    '''

    prompt=f'''
    Generate one food recepie using these ingredients: {','.join(ingredients)}
    Recepie should not be more than 100 words
    Cuisine:{cuisine}
    Diet:{diet}
    '''
    response=client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )
    return response.text
```
