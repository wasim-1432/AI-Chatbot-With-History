```python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client=genai.Client()
```

```python
response=client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Hi! My name is Mohd Wasim. Please remember it."
)
print(response.text)
    
```

```python
response=client.models.generate_content(
    model="gemini-3.6-flash",
    contents="What is my name?"
)
print(response.text)
```

```python
conversation=[]
def chat(user_message):
    conversation.append({"role":"user","parts":[{"text":user_message}]})
    try:
        response=client.models.generate_content(
        model="gemini-3.6-flash",
        contents=conversation
        )
        reply=response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"
    conversation.append({"role":"model","parts":[{"text":reply}]})
    return reply
```

```python
print(chat("Hi! My name is Mohd Wasim. Please remember it."))
```

```python
print(chat("What is my name?"))
```

```python
for message in conversation:
    print(message["role"],"->",message["parts"][0]["text"])
```
